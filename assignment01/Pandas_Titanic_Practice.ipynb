- Cabin : 객실 번호

```python
del df['Cabin'] # Cabin 열 전체 삭제
df = df.dropna() # 결측치 (NaN)가 하나라도 있는 행을 제거한 결과를 다시 df에 저장
df = df.reset_index(drop=True)  # 인덱스를 0부터 다시 지정
df
```

## 3) 타이타닉 데이터 프레임 Human_classfi 컬럼 추가하기
- 남자이며, 19세 미만은 [Child_m]으로 설정
- 여자이며, 19세 미만은 [Child_f]으로 설정
- 남자이며, 19세 이상은 [Mr]으로 설정
- 여자이며, 19세 이상은 [Ms]으로 설정

'''
    TODO:: Human_classfi 컬럼 추가 (위 조건에 맞는 값으로 넣을것)
    Hint : loc 함수 사용, 다중 조건 사용
'''

```python
df['Age'].dtype
```

```python
df.loc[(df['Sex'] == 'male') & (df['Age'] < 19), 'Human_classfi']= 'Child_m'
df.loc[(df['Sex'] == 'female') & (df['Age'] < 19), 'Human_classfi'] = 'Child_f'
df.loc[(df['Sex'] == 'male') & (df['Age'] >= 19), 'Human_classfi'] = 'Mr'
df.loc[(df['Sex'] == 'female') & (df['Age'] >= 19), 'Human_classfi'] = 'Ms'

df
```

## 4) 아래의 데이터 프레임을 타이타닉 데이터 프레임에 조인하기

```python
data_dict = {
    "Human_classfi" : ['Child_m', 'Child_f', 'Mr', 'Ms'],
    "HC_DES" : ['19세 미만의 남성', '19세 미만의 여성','19세 이상의 남성','19세 이상의 여성']
}

des_df = pd.DataFrame(data_dict)
des_df
```

```python
pd.merge(df, des_df, on='Human_classfi', how='left')
```
