## **3. 인덱스 활용**
### **1) 1행 10열의 배열을 랜덤으로 5개 생성**

```python
arr = [np.random.rand(1, 10) for _ in range(5)]
arr
```

```python
np.random.rand(5, 1, 10)
```

```python
arr1 = np.random.randint(1, 10, size=10)
arr2 = np.random.randint(1, 10, size=10)
arr3 = np.random.randint(1, 10, size=10)
arr4 = np.random.randint(1, 10, size=10)
arr5 = np.random.randint(1, 10, size=10)
```

### **2) 각 배열을 vstack을 이용하여 합치기**

```python
v_arr = np.vstack([arr1, arr2, arr3, arr4, arr5])
v_arr
```

### **3) 5행 10열의 배열을 랜덤으로 생성 후, 2)에서 합친 배열과 수평축으로 합치기**

```python
arr_new = np.random.randint(1, 10, size=(5,10))
arr_new
```

```python
arr_h = np.hstack([v_arr, arr_new])
arr_h
```

### **4) 3)을 전치**

```python
arr_T = arr_h.T
arr_T
```

### **2) 특정 열 인덱스를 기준으로 행을 정렬**

```python
idx = np.argsort(arr_T[:,0])
idx
a = arr_T[idx]
a
```

## **4. Random.Choice 활용**

### **1) 5행 6열 배열을 랜덤으로 생성**

- 각 행의 원소는 1~45사이의 랜덤 값이다.
- 각 행의 원소들은 행 별로 중복되지 않은 수로 구성한다.
- 1행 6열 배열을 5개 만들어 합치도록 코드를 작성하라.

```python
arr1 = np.random.choice(np.arange(1, 46), 6, replace=False).reshape(1, 6)
arr2 = np.random.choice(np.arange(1, 46), 6, replace=False).reshape(1, 6)
arr3 = np.random.choice(np.arange(1, 46), 6, replace=False).reshape(1, 6)
arr4 = np.random.choice(np.arange(1, 46), 6, replace=False).reshape(1, 6)
arr5 = np.random.choice(np.arange(1, 46), 6, replace=False).reshape(1, 6)

arr_v = np.vstack([arr1, arr2, arr3, arr4, arr5])
arr_v
```

### **2) 1~45의 수를 랜덤으로 6개를 뽑는 코드를 작성**
- 단, 원하는 숫자 6개를 원하는 확률을 부여하여 생성할 것.

```python
nums = np.arange(1, 46)
nums
```

```python
p = np.zeros(45) # 모든 숫자 기본 가중치 0
p
```

```python
p[[2, 6, 11, 19, 32, 44]] = 0.16666667 # p의 특정 위치만 가중치 5로 지정
p
```

```python
result = np.random.choice(nums, 6, p=p/p.sum())
result
```
